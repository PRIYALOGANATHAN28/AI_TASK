```python

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report




```

```python
spam_msgs = [
    "Win cash prize now!!!",
    "Congratulations! You won a lottery",
    "Get free recharge today",
    "Urgent! Your bank account is blocked",
    "You have won 5,00,000 rupees",
    "Limited offer, click the link",
    "Buy now and get 80% off",
    "Your loan is approved",
    "Claim your free voucher now",
    "Exclusive offer only for today",
    "Free vacation to Goa",
    "You are selected for a prize",
]

ham_msgs = [
    "Hey, where are you?",
    "Call me when you're free",
    "Let's meet tomorrow",
    "Send me the documents",
    "I reached home safely",
    "Don't forget the meeting",
    "Please check your email",
    "What time is the class?",
    "Thank you so much",
    "I'll come at 6 PM",
    "Message me after lunch",
    "Did you finish homework?",
]
```

```python

train_df, test_df = train_test_split(df, test_size=0.20, random_state=42, shuffle=True)

# Print sizes
print("After 80/20 split:")
print("Train:", len(train_df))
print("Test:", len(test_df))
```

```python
# FIX: Rectify sampling errors by checking size first

# For training data
if len(train_df) >= 150:
    train_df = train_df.sample(150, random_state=10)
else:
    print(f"train_df has only {len(train_df)} rows, cannot sample 150.")
    train_df = train_df.sample(len(train_df), random_state=10)  # sample all available

# For testing data
if len(test_df) >= 50:
    test_df = test_df.sample(50, random_state=20)
else:
    print(f"test_df has only {len(test_df)} rows, cannot sample 50.")
    test_df = test_df.sample(len(test_df), random_state=20)  # sample all available

print("\nFinal Dataset:")
print("Train =", len(train_df))
print("Test  =", len(test_df))
```

```python
# --------------------------------------------------------------
# 5. TF-IDF Vectorization
# --------------------------------------------------------------

vectorizer = TfidfVectorizer()
X_train = vectorizer.fit_transform(train_df["text"])
X_test = vectorizer.transform(test_df["text"])

y_train = train_df["label"]
y_test = test_df["label"]
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)

# Predictions
pred = model.predict(X_test)
```

```python
print("\nAccuracy:", accuracy_score(y_test, pred))
print("\nClassification Report:\n", classification_report(y_test, pred))

# Probabilities
print("\nSample Probabilities (first 10):")
print(model.predict_proba(X_test)[:10])
```
